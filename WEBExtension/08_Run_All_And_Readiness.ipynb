Configure adequate costs before calling all phases. E2 exact prompts reuse E1 R, so they should create no additional physical calls. Separate experiment checkpoints remain available for interpretation.

A full 300-question default suite with E1 self-updates, E3 and depth-3 E4 has a conservative upper bound of 11,400 calls before cache reuse (E1 7,200 + E3 600 + E4 3,600; E2 reuses E1). Some E4 initial/first-update prompts also match E1. These are completions, not prices. Seeds multiply cost. Do not call a run complete until all intended questions, strict invariants, raw responses and scorer outputs validate.
